# Structured output, providing schema to a chat model

```python
import os
from langchain_groq import ChatGroq
from pydantic import Field, BaseModel
from dotenv import load_dotenv
load_dotenv()
```

```python
groq = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
raw_response = groq.invoke("Provide details of blockbuster Ramayan serial")
print(raw_response.content)
```

```python
class Actor(BaseModel):
    name: str = Field(description="Name of the actor")

class Movie(BaseModel):
    name: str = Field(description="Name of the movie or tv show")
    release_date: str = Field(description="Date of release in dd/mm/yyyy format")
    genere: str = Field(description="Genere of the movie or tv show")
    brief: str = Field(description="Brief of the moview or show within 25 words")
    actors: list[Actor]



groq_with_structure = groq.with_structured_output(Movie, include_raw=True)
response = groq_with_structure.invoke("Provide details of blockbuster Ramayan serial")
response
```

### Typed Dict
Use it when you don't need pydantic validation or runtime validation.

```python
from typing_extensions import TypedDict

# If instead of BaseModel class, if we use TypedDict
# No type validation happens, just the structure
class Actor(TypedDict):
    name: str = Field(description="Name of the actor")

class Movie(TypedDict):
    name: str = Field(description="Name of the movie or tv show")
    release_date: str = Field(description="Date of release in dd/mm/yyyy format")
    genere: str = Field(description="Genere of the movie or tv show")
    brief: str = Field(description="Brief of the moview or show within 25 words")
    actors: list[Actor]



groq_with_structure = groq.with_structured_output(Movie, include_raw=True)
response = groq_with_structure.invoke("Provide details of blockbuster Ramayan serial")
response
```
