### Model integration
langchain can be used to integrate any LLM models like `OpenAI`, `Groq`, `Gemini`.
`langchain.chat_model.init_chat_model`: can be used to integrate, provide model names like `gpt-5` or `gpt-4.1` to talk to OpenAI models. 
2. For Groq use `groq:` in the model name
3. For google Gemini use `google_gemini`

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
groqGpt = init_chat_model(model="groq:openai/gpt-oss-120b")
groqGpt
```

```python
response = groqGpt.invoke("What is the latest news about NVIDIA?")
```

```python
print(response.content)
```

```python
from langchain_google_genai.chat_models import ChatGoogleGenerativeAI
```

```python
googleGemini = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = googleGemini.invoke("What is the latest news about NVIDIA?")
response.content
```
